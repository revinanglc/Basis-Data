```python
!mysql -u root -D toko_matematika_ceria -e "CREATE TABLE transaksiii (id_transaksi INT AUTO_INCREMENT PRIMARY KEY, id_produk INT, jumlah INT NOT NULL, total_harga DECIMAL(10,2), FOREIGN KEY (id_produk) REFERENCES produk(id_produk));"
```

```python
!mysql -u root -D toko_matematika_ceria -e "INSERT INTO transaksiii (id_produk, jumlah, total_harga) SELECT id_produk, 2, harga*2 FROM produk WHERE id_produk=1; INSERT INTO transaksiii (id_produk, jumlah, total_harga) SELECT id_produk, 1, harga*1 FROM produk WHERE id_produk=3; INSERT INTO transaksiii (id_produk, jumlah, total_harga) SELECT id_produk, 3, harga*3 FROM produk WHERE id_produk=5; INSERT INTO transaksiii (id_produk, jumlah, total_harga) SELECT id_produk, 2, harga*2 FROM produk WHERE id_produk=6;"
```

#Tabel transaksi yang mencatat pembelian

```python
!mysql -u root -D toko_matematika_ceria -e "SELECT * FROM transaksiii;"
```

#Total penjualan per produk

```python
!mysql -u root -D toko_matematika_ceria -e "SELECT p.nama_produk, SUM(t.jumlah) AS total_terjual, SUM(t.total_harga) AS total_pendapatan FROM transaksiii t JOIN produk p ON t.id_produk = p.id_produk GROUP BY p.nama_produk;"
```
