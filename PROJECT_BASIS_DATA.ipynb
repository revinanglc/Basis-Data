```python
conn = sqlite3.connect("ecommerce_indonesia.db")
cursor = conn.cursor()
```

#FUNGSI MENJALANKAN QUERY

```python
def run_query_and_display(query_name, sql_query):

    print(f"\n{'=' * 70}")
    print(f"QUERY: {query_name}")
    print(f"{'=' * 70}")

    print(f"SQL:\n{sql_query}")

    try:
        result = pd.read_sql_query(sql_query, conn)

        print("\nHasil:")
        print(result.to_string(index=False))

    except Exception as e:
        print("ERROR:", e)
```

# 1. TOTAL PEMBAYARAN TERBESAR

```python

run_query_and_display(
    "Top 5 Total Pembayaran Terbesar",

    """
    SELECT
        order_id,
        product_categories,
        `Metode Pembayaran`,
        `Total Pembayaran`
    FROM ecommerce
    ORDER BY `Total Pembayaran` DESC
    LIMIT 5;
    """
)
```

# 2. JUMLAH ORDER PER PROVINSI

```python
run_query_and_display(
    "Jumlah Order per Provinsi",

    """
    SELECT
        Provinsi,
        COUNT(order_id) AS jumlah_order
    FROM ecommerce
    GROUP BY Provinsi
    ORDER BY jumlah_order DESC;
    """
)
```

# 3. TOTAL PENJUALAN PER KATEGORI

```python
run_query_and_display(
    "Total Penjualan per Kategori",

    """
    SELECT
        product_categories,
        SUM(`Total Pembayaran`) AS total_penjualan
    FROM ecommerce
    GROUP BY product_categories
    ORDER BY total_penjualan DESC
    LIMIT 10;
    """
)
```

# 4. METODE PEMBAYARAN PALING SERING DIGUNAKAN

```python
run_query_and_display(
    "Metode Pembayaran Terpopuler",

    """
    SELECT
        `Metode Pembayaran`,
        COUNT(*) AS total_penggunaan
    FROM ecommerce
    GROUP BY `Metode Pembayaran`
    ORDER BY total_penggunaan DESC;
    """
)
```

# 5. ORDER DENGAN ONGKIR TERTINGGI

```python
run_query_and_display(
    "Top Ongkir Tertinggi",

    """
    SELECT
        order_id,
        Provinsi,
        `Perkiraan Ongkos Kirim`
    FROM ecommerce
    ORDER BY `Perkiraan Ongkos Kirim` DESC
    LIMIT 10;
    """
)
```

# 6. STATUS PESANAN

```python
run_query_and_display(
    "Jumlah Status Pesanan",

    """
    SELECT
        `Status Pesanan`,
        COUNT(*) AS jumlah
    FROM ecommerce
    GROUP BY `Status Pesanan`;
    """
)
```
